```python
plt.figure(figsize=(12,6))

sns.barplot(data=df,x="workclass",y="-age",hue="income")
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    pd.crosstab(
        df["occupation"],
        df["race"],
        normalize="columns"
    ) * 100,
    annot=True,
    fmt=".1f",
    cmap="magma"
)

plt.xlabel("Race")
plt.ylabel("Occupation")
plt.title("Occupational Distribution Within Each Race (%)")
plt.show()
```
